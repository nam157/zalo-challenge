```python
import cv2
import os
import random
root_dir = "/home/namnh1/zalo-challenge/video-train/"
output_dir = "/home/namnh1/zalo-challenge/data/1"

# Ensure the output directory exists
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
ls = []
for video_name in os.listdir(root_dir):
    video_path = os.path.join(root_dir, video_name)
    cap = cv2.VideoCapture(video_path)
    i = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break  # Exit the loop if no more frames are available

        # Save every 10th frame
        if i % 10 == 0:
            ims = random.randint(1100,5000)
            output_image_path = os.path.join(output_dir, f"image_real_{ims}.jpg")
            cv2.imwrite(output_image_path, frame)
            ls.append([output_image_path,1])
        i += 1

    cap.release()
```
